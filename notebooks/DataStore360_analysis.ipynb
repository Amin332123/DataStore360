```python
# find missing values and outliers in the ship date and order data . 


dataset["Ship Date"].describe()

dataset[["Ship Date" , "Order Date"]]

# i want to see if there a relation between the date of ordering and date of shipping .
# if yes there i can replace the missing values with a data that depends on my calculations .

# i will get the days while forgetting the months . and see th amount of days that the shipping time is taking . 

# i fix all the dates formats to one strucutred format : year/month/day
dataset["Order Date"] = pd.to_datetime(
    dataset["Order Date"],
    format="mixed"
)

dataset["Ship Date"] = pd.to_datetime(
    dataset["Ship Date"],
    format="mixed"
)

dataset["shipping_days"] = (dataset["Ship Date"] - dataset["Order Date"]).dt.days

dataset
dataset[["Ship Date", "Order Date", "shipping_days"]]

# now i know how many days each order take to get delieved to the customer . 
#print(f"shiiping days : {(dataset["Ship Date"] - dataset["Order Date"]).dt.days}")


# i will see if there negative values : 

dataset.loc[dataset["shipping_days"] < 0, "shipping_days"]

# I saw in the rapport that there are missing shipping modes so i filled them with the mode of the shipping mode 
dataset["Ship Mode"] = dataset["Ship Mode"].fillna(
    dataset["Ship Mode"].mode()[0]
)

# Here we get only the validated shipping days (>= 0) .
valid = dataset["shipping_days"] >= 0

# Here i only get the ship Mode name and its mode . 
mode_days = (
    dataset[valid]
    .groupby("Ship Mode")["shipping_days"]
    .agg(lambda x: x.mode().iloc[0])
)

# Here i do two condition to get nan ship Dates and also incorrect ship dates
condition = (
    dataset["Ship Date"].isna() |
    (dataset["Ship Date"] < dataset["Order Date"])
)

# Here i take the order date and i add to it number of days that is compatible with the shipping mode . 
dataset.loc[condition, "Ship Date"] = (
    dataset.loc[condition, "Order Date"]
    + pd.to_timedelta(
        dataset.loc[condition, "Ship Mode"].map(mode_days),
        unit="D"
    )
)

 
# Recalculate shipping_days after correcting Ship Date
dataset["shipping_days"] = (
    dataset["Ship Date"] - dataset["Order Date"]
).dt.days

# displaying only to see if there are missing columns and their percentage . 
for column in dataset.columns :
    nullValues = dataset[column].isna() 
    print(f"{column} => {nullValues.sum()} , and {(nullValues.mean() * 100):.2f}%")




```

```python
# this part i will work on Customer Name because i saw we have 351 missing values there . 

# small investigation . to see each customer id and how many times his name is repeated . 

# i see like the not unique customer name count and grouped by their ids . 
dataset.groupby("Customer ID")["Customer Name"].nunique().value_counts()



# I make names from this "  Amine rah" to "Amine Rah" . 
dataset["Customer Name"] = dataset["Customer Name"].str.strip().str.title()

# I fill the nan values of customer names with their real names . 
dataset["Customer Name"] = dataset["Customer Name"].fillna(
    dataset.groupby("Customer ID")["Customer Name"]
    .transform("first")
)



# For the person remaining with no customer name . we will put his name same as his id so later it will be hashed . 
dataset["Customer Name"] = dataset["Customer Name"].fillna(
    dataset["Customer ID"]
)

# checking if everything went correct to move to next step . 
for column in dataset.columns :
    nullValues = dataset[column].isna() 
    print(f"{column} => {nullValues.sum()} , and {(nullValues.mean() * 100):.2f}%")




```

```python
# in this part i will work on postal code . 

dataset.groupby("State")["Postal Code"].nunique().value_counts()



dataset.loc[
    dataset["Postal Code"].isna(),
    ["City", "State", "Postal Code"]
]


# i get here only postal modes . 






""" in these lines of codes , i take each city with its mode postal code and also 
each state with its postal code and then assign the postal code to the ones who have nan values 
"""
city_postal_codes = dataset.groupby("City")["Postal Code"].agg(
    lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan
)

state_postal_codes = dataset.groupby("State")["Postal Code"].agg(
    lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan
)

dataset["Postal Code"] = dataset["Postal Code"].fillna(
    dataset["City"].map(city_postal_codes)
)


dataset["Postal Code"] = dataset["Postal Code"].fillna(
    dataset["State"].map(state_postal_codes)
)



# checking if everything went correct to move to next step . 
for column in dataset.columns :
    nullValues = dataset[column].isna() 
    print(f"{column} => {nullValues.sum()} , and {(nullValues.mean() * 100):.2f}%")
```
